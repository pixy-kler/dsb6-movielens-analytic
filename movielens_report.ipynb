```python
%%timeit -n 1 -r 1 -q
from movielens_analysis import Movies, Ratings, Tags, Links

movies = Movies('ml-latest-small/movies.csv')
ratings = Ratings('ml-latest-small/ratings.csv')
tags = Tags('ml-latest-small/tags.csv')
links = Links('ml-latest-small/links.csv')

globals()['movies'] = movies
globals()['ratings'] = ratings
globals()['tags'] = tags
globals()['links'] = links

print("\n")
print(f"\033[1mАКТ ПЕРВЫЙ. ВХОД В ТЕМНОТУ\033[0m")
print("\n")
```

```python
%%timeit -n 1 -r 1 -q
movies = globals().get('movies')
stats = movies.get_summary_stats()

print("\n")
print(f"Софит освещает плакат на стене:")
print("\n")
print(f"Человек в ковбойской шляпе стоит, вытянув вперёд руку с отведённым вверх большим пальцем.")
print(f"Надпись ниже: Мы продемонстрируем вам пример анализа статистики с сайта MovieLens на выборке в 1000 фильмов!")
print("\n")
print(f"Среди них дата указана у {stats['movies_with_year']} фильма, что говорит о большом объёме работ - ребята реально работают с Big Data!")
print("\n")
print(f"\033[1mГоворят, самый старый фильм в истории кинематографа датируется {stats['oldest_movie']} годом.\033[0m")
print("На самом сайте, впрочем, поискать по годам для интереса не получится - всё только через парсинг, детка!")
print("\n")
print("Зато легко можно определить любимые жанры пользователей сайта:")
for genre, count in stats['top_genres']:
    print(f"  {genre} — {count} фильмов")
print("\n")
print("Гаснет свет.")
print("\n")
print(f"\033[1mАКТ ВТОРОЙ. ГОЛОСА ИЗ ТЕМНОТЫ\033[0m")
print("\n")
```


Оживает экран: на нём появляются кадры. Зрители смотрят и перешёптываются.
    

```python
%%timeit -n 1 -r 1 -q
ratings = globals().get('ratings')
movies = globals().get('movies')

stats = ratings.get_stats()
top_movies = ratings.get_top_movies(n=10)
movie_titles = {m['movieId']: m['title'] for m in movies.movies}

print("\n")
print("ЧТО МЫ ШЕПТАЛИ В ТЕМНОТЕ")
print(f"Сегодня для анализа мы взяли {stats['total']} фильмов, оцененных зрителями;")
print(f"При этом {stats['unique_users']} из этих зрителей считаются уникальными;")
print(f"А среди фильмов таковыми считаются {stats['unique_movies']};")
print(f"Средняя оценка: {stats['avg_rating']};")
print(f"Разброс оценок - от {stats['min_rating']} до {stats['max_rating']}.")
print("\n")
print(f"\033[1mТОП-10 фильмов по мнению пользователей сайта MovieLens\033[0m")
for i, (movie_id, weighted_rating, avg_rating, rating_count) in enumerate(top_movies, 1):
    title = movie_titles.get(movie_id, 'Неизвестный фильм')
    print(f"{i}. {avg_rating:.2f} — {title}")
print("\n")
```


В реальной критике фильмов с идеальным средним баллом практически не бывает. 
Здесь же мы наблюдаем идеальный ТОП-10. Это говорит о двух вещах:

1) Самоотбор аудитории: MovieLens любят киноманы старой закалки, которые ценят сценарий, харизму и практические эффекты. Они не накручивают рейтинг новинкам.

2) Эмоциональный максимум: Пользователь ставит 5 не за «объективное величие» «Гражданина Кейна», а за фильм, который он пересматривал 20 раз и который сформировал его вкус.

**Три столпа этого пантеона**

А. Архетипическое приключение (№1, 3, 7, 8, 9)
«Звездные войны. Новая надежда», «Волшебник из страны Оз», «Принцесса-невеста», «Терминатор», «Назад в будущее».
Это идеальные врата в мир кино. Каждый из них — безупречный образец своего жанра: космическая опера, фэнтези, сказка для взрослых, техно-нуар и научная фантастика. Их объединяет универсальный миф (путь героя, спасение мира, любовь, побеждающая время). Они учат нас, что кино — это чистое удовольствие без чувства вины.

Б. Абсурд и подрыв (№5, 6)
Два фильма Монти Пайтона — «Житие Брайана» и «Священный Грааль». Их присутствие рядом с «Фантазией» — гениальный контрапункт. Аудитория MovieLens обожает умный, литературный юмор, граничащий с философией. «Жизнь Брайана» — это не просто комедия, а манифест свободы от догм. Тот факт, что он стоит выше «Терминатора», говорит о зрелости аудитории.

В. Странные включения (№2, 4)
Вот где начинается магия краудсорсинга.

«Беглец» (1993) с Харрисоном Фордом — отличный триллер, но 4.75 и второе место в истории? Это явно результат «эффекта cult following». Вероятно, это фильм, который идеально работает в своей нише: крепкий, умный мейнстрим, где нет слабых мест.

«Призрак и Тьма» (1996) — настоящий шокер. Фильм про львов-людоедов с Вэлом Килмером. Для большинства критиков это проходной приключенческий хоррор, а здесь он в пятерке величайших. Вывод: на MovieLens живут фанаты атмосферного исторического экшена и Вэла Килмера в ударе.

Единственный фильм, который выбивается из нарративного кино - это «Фантазия» (1940). Это чистый визуал, музыка и эксперимент. Его высокий балл доказывает, что пользователи MovieLens ценят амбицию и уникальность больше, чем связный сюжет. Это кино как искусство, а не как история.


```python
%%timeit -n 1 -r 1 -q
tags = globals().get('tags')

stats = tags.get_stats()
popular_tags = tags.get_most_popular_tags(n=10)

print("\n")
print("СЛОВА, КОТОРЫЕ МЫ ШЕПТАЛИ")
print(f"Пользователи сайта придумали {stats['total_tags']} тегов, описывающих фильмы.")
print(f"Стоит ли произносить {stats['unique_users']} имён, создавших наиболее заковыристые из них?")
print("\n")
print(f"Мы готовы назвать самые популярные теги.")
print("\n")
print(f"\033[1mТЫ ЖДАЛ ЭТОГО.\033[0m")
print("\n")
for tag, count in popular_tags[:5]:
    print(f"  «{tag}» — {count} раз")
print("\n")
```

15 упоминаний «funny» + 12 «dark comedy» - это почти половина всей теговой активности пользователей сайта.

**Что это значит?**

Аудитория MovieLens не просто любит смеяться — она пользуется юмором как навигационным инструментом. Для них «смешной» — это не жанр, а гарантия качества: фильм не будет скучным, пафосным или медленным.

А «dark comedy» уточняет: смех должен быть с горчинкой. Это объясняет высокие позиции «Жития Брайана» и «Священного Грааля» — там смеются над смертью, верой, властью и собственным телом. Такой юмор считается «умным» и «взрослым».

Обратите внимание на тег «twist ending» — это тег-предупреждение. Он говорит: «Не читай синопсис, не смотри трейлер, просто смотри». В эпоху стриминга, где всё рассказывают до просмотра, этот тег становится актом сопротивления. Это договор: я доверяю тебе, фильм, и ты меня не подведешь в последние десять минут.

Обратите внимание, чего нет в списке:

«романтика» - 0 раз; 
«мелодрама» - 0 раз; 
«депрессивно» - 0 раз; 
«биография» - 0 раз.

Это полное отрицание того, что обычно приносит Оскары и слёзы критиков. Аудитория MovieLens избегает эмоционального дискомфорта. Им не нужен «Список Шиндлера» или «Мальчик в полосатой пижаме». Они приходят в кино за удовольствием, интеллектуальной игрой и контролируемым напряжением.


```python
%%timeit -n 1 -r 1 -q
links = globals().get('links')
tags = globals().get('tags')

links_stats = links.get_stats()
tags_stats = tags.get_stats()
without_tmdb = links.get_movies_without_tmdb()
missing = links.find_missing_ids()
toy = links.get_by_movie_id(1)

total_tags = tags_stats['total_tags']

with_tmdb = links_stats['with_tmdb']
without_tmdb_count = len(without_tmdb)

missing_count = len(missing)

print("\n")
print("Внезапно экран вспыхивает - и гаснет. Снова темнота. Голоса постепенно стихли.")
print("Тут раздались шаги.")
print("\n")
print("=" * 70)
print("\n")
print(f"Нам понадобится большая и дружная команда, чтобы вместе дописать эту историю:")
print(f"для описания собственных ощущений пользователи нагенерили уже {total_tags} тегов.")
print(f"А котируются в tmdb {with_tmdb} фильмов!")
print(f"Или хотя бы найти вот эти {without_tmdb_count} 'нечипированных' фильма. Здесь точно нужна командная работа!")
print("\n")
print(f"А о существовании скольких авторских и национальных фильмов MovieLens и не догадывается!")
print("\n")
print(f"Кстати, есть тут и {missing_count} случаев, в которых требуется вмешательство дотошного редактора.")
if toy:
    imdb_clean = toy['imdbId'].lstrip('0')
    print(f"Например здесь: Toy Story на IMDb — https://www.imdb.com/title/tt{imdb_clean}/")
print("\n")
```

```python
%%timeit -n 1 -r 1 -q
movies = globals().get('movies')
ratings = globals().get('ratings')
tags = globals().get('tags')
links = globals().get('links')

movies_stats = movies.get_summary_stats()
ratings_stats = ratings.get_stats()
tags_stats = tags.get_stats()
links_stats = links.get_stats()

print("\n")
print(f"\033[1mИТОГОВАЯ СТАТИСТИКА:\033[0m")
print(f"   • {movies_stats['total_movies']} фильмов прошли перед нашими глазами. От абсурда Монти Пайтона до совершенной механики «Назад в будущее».")
print(f"   • {ratings_stats['total']} оценок прозвучало в темноте — каждая из которых могла быть импульсивной «пятёркой» после титров или выстраданной «тройкой» на следующий день.")
print(f"   • {tags_stats['total_tags']} слов, прошёптанных в темноте кинозала: «Смотри, сейчас будет твист», «Опять этот скучный диалог», «Господи, как же я люблю это кино». ")
print(f"   • {links_stats['total']} связей с внешним миром - потому что ни один просмотр не случается в вакууме. Мы смотрим с кем-то, спорим после, ставим теги «funny» или «dark comedy», рекомендуем друзьям или предупреждаем: «Только не говори, чем кончается».")
print(f"   • {movies_stats['movies_with_year']} лет киноистории мы пережили заново.")
print("\n")
```

Данные, которые мы разложили по полочкам — рейтинги, теги, частотности, — это не холодная аналитика. Это термометр коллективной души. Когда 15 человек ставят тег «sci-fi» и 15 — «funny», они не классифицируют фильм. Они говорят: «Вот что делает меня счастливым. Вот воздух, которым я хочу дышать в темноте».

За каждым рейтингом — человек, который смеялся или плакал. Который в одиночестве или в обнимку с попкорном прожил два часа чужой жизни и сделал её своей. Который после финальных титров «Жития Брайана» вышел из зала чуть более свободным, чем был до этого.
